# 7. Train-Test Split and Data Preprocessing

Before training Machine Learning models, the dataset is prepared by encoding the target variable, separating the features and target, and splitting the data into training and testing sets.

The training data is used to train the models, while the testing data is used to evaluate performance on unseen data.

Target Encoding

```python
print(df["Heart Disease"].value_counts(dropna=False))
```

```python
df["Heart Disease"] = df["Heart Disease"].map({
    "Absence": 0,
    "Presence": 1
})
```

```python
print("Target column:")
print(df["Heart Disease"].head())

print("\nTarget data type:")
print(df["Heart Disease"].dtype)

print("\nTarget value counts:")
print(df["Heart Disease"].value_counts())
```

```python
print(df["Heart Disease"].value_counts(dropna=False))

print(
    "Missing values:",
    df["Heart Disease"].isna().sum()
)
```

## Separate X and y

The target column `Heart Disease` is separated from the input features.

- `X` contains the 13 input features.
- `y` contains the target variable.

```python
X = df.drop("Heart Disease", axis=1)

y = df["Heart Disease"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTraining data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

Data Preprocessing

# 8. Data Preprocessing

### Encoding

The categorical target variable `Heart Disease` contains the classes `Absence` and `Presence`.

These classes are encoded as:

- `Absence` → 0
- `Presence` → 1

The input features in this dataset are already represented using numerical codes, so separate feature encoding is not required.

## Pipeline

A Machine Learning pipeline is used to combine preprocessing and model training into a single workflow.

`StandardScaler` is used to standardize numerical features before Logistic Regression.

```python
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=2000))
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)

y_prob = pipeline.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import accuracy_score

y_train_pred = pipeline.predict(X_train)
train_accuracy = accuracy_score(y_train, y_train_pred)

y_test_pred = pipeline.predict(X_test)
test_accuracy = accuracy_score(y_test, y_test_pred)

print("Training Accuracy:", train_accuracy)
print("Testing Accuracy:", test_accuracy)
```

# 9. Handling Imbalance

The class distribution of the target variable is checked to determine whether oversampling or undersampling is required.

```python
print(y.value_counts())

print(
    y.value_counts(normalize=True) * 100
)
```

Therefore, no oversampling or undersampling technique was required for this dataset.
