# 14. Cross Validation

Cross-validation is used to obtain a more reliable estimate of model performance.

Stratified K-Fold Cross Validation is used because this is a classification problem. It maintains the class distribution across the different folds.

```python
df["Heart Disease"] = df["Heart Disease"].map({
    "Absence": 0,
    "Presence": 1
})
```

```python
X = df.drop("Heart Disease", axis=1)
y = df["Heart Disease"]
```

```python
logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=2000))
])
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scores = cross_val_score(
    logistic_model,
    X,
    y,
    cv=cv,
    scoring="accuracy"
)

print("Cross-validation scores:", scores)

print(
    "Mean CV Accuracy:",
    scores.mean()
)
```

### Interpretation

The mean cross-validation accuracy provides an overall estimate of how consistently the Logistic Regression model performs across different subsets of the dataset.
